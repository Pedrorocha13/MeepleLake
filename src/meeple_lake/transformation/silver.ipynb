```python
import xml.etree.ElementTree as ET
```

```python
tree = ET.parse('/home/prjesus/projects/MeepleLake/data/bronze/bgg/thing/game_174430_2026-08-28.xml')
root = tree.getroot()

print("Root element:", root.tag)
```

```python
for child in root:
    print("Child element:", child.tag, "with attributes:", child.attrib)
```

```python
item = root.find("item")
game_id = item.get("id")
primary_name = item.find("name[@type='primary']")
game_name = primary_name.get("value")

print(game_id)
print(game_name)
```

```python
data_dict = {
    "id": game_id,
    "name": game_name
}

print(data_dict)
```

```python
import pandas as pd
```

```python
dados_list = [data_dict]
df = pd.DataFrame(dados_list)
print(df)
```
